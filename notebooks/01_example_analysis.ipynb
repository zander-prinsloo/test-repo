# Example Data Analysis

This notebook demonstrates a typical data analysis workflow.

## 1. Setup

Import necessary libraries and configure settings.

```python
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 2. Load Data

Load the dataset from the data folder.

```python
# Example: Load data using utility function
# from utils import load_data
# df = load_data('../data/raw/your_data.csv')

# For demonstration, create sample data
df = pd.DataFrame({
    'date': pd.date_range('2023-01-01', periods=100),
    'value': np.random.randn(100).cumsum(),
    'category': np.random.choice(['A', 'B', 'C'], 100)
})

df.head()
```

## 3. Exploratory Data Analysis

Understand the data through summary statistics and visualizations.

```python
# Summary statistics
df.describe()
```

```python
# Visualize data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(df['date'], df['value'])
plt.title('Time Series Plot')
plt.xlabel('Date')
plt.ylabel('Value')

plt.subplot(1, 2, 2)
df.groupby('category')['value'].mean().plot(kind='bar')
plt.title('Average Value by Category')
plt.xlabel('Category')
plt.ylabel('Average Value')

plt.tight_layout()
plt.show()
```

## 4. Data Processing

Clean and transform the data as needed.

```python
# Example processing steps
df_processed = df.copy()

# Add derived features
df_processed['rolling_mean'] = df_processed['value'].rolling(window=7).mean()

df_processed.head(10)
```

## 5. Analysis and Insights

Perform analysis and document findings.

```python
# Analysis code here
print("Key findings:")
print(f"- Total observations: {len(df)}")
print(f"- Categories: {df['category'].unique()}")
print(f"- Value range: {df['value'].min():.2f} to {df['value'].max():.2f}")
```
